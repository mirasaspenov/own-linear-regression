I am gonna make my own linear regression to understand what is happening under the hood and compare it with sklearn's model.
In first, lets import all packages and create a data

```python
#importing all necessary libriaries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.datasets import make_regression
%matplotlib inline
import warnings
warnings.filterwarnings("ignore")
```

```python
#loading the data (one feature)
X, y = make_regression(n_samples=100, n_features=1, n_informative=1, random_state=18, noise=0.5)
```

```python
display(X,y)
```

```python
plt.scatter(X,y)
plt.xlabel("Feature")
plt.ylabel("Target")
```

```python
#training the sklearn model
reg = LinearRegression()
reg.fit(X, y)
```

```python
reg.coef_, reg.intercept_
```

```python
plt.scatter(X,y)
plt.plot(X,reg.predict(X), color='green', label="Sklearn's Linear Regression")
plt.xlabel("Feature")
plt.ylabel("Target")
plt.legend()
```

Here we have created sklearn model and displayed its parameters. But how did the model got this numbers? Why we just can not use our own numbers ? Let's say -3 and 2

```python
plt.scatter(X,y)
plt.plot(X,-3*X[:,0] + 2, color='green', label="Our Linear Regression")
plt.legend()
```

It is clear that this model would work much worse, in order to find appropriate values we need some optimization. But first, we need to somehow measure the accuracy of the model, then the MSE comes in. For better understanding lets create its graph

```python
import mpl_toolkits.mplot3d.axes3d as p3

def mse(w1,w0):
    y_pred = w1*X[:,0] + w0
    return np.mean((y_pred-y)**2)

fig = plt.figure(figsize=[13,13])

lin1 = np.linspace(-4,12,100)
lin2 = np.linspace(-4,4,100)

w1,w0 = np.meshgrid(lin1,lin2)

z = np.array([mse(i,j) for i,j in zip(np.ravel(w1),np.ravel(w0))])
Z = z.reshape(w1.shape)

ax = plt.axes(projection='3d')

ax.plot_surface(w1, w0, Z, label='MSE function', alpha=0.5)
ax.set_xlabel('w1')
ax.set_ylabel('w0')
ax.scatter(reg.coef_, reg.intercept_, color='orange', label="Sklearn's parameters",  )
ax.scatter(-3, 2, color='blue', label="Our parameters" )
plt.legend()

```

The goal is to have the lowest mse, the sklearn's point is at the lowest as we see, but the blue one which is ours, is too far from the bottom

So we need to find the lowest MSE, for this purpose we already have the gradient descent, which would be ideal for us. So let's optimize it!

```python
def mse(X,y,w1,w0):
    y_pred = w1*X[:,0] + w0
    return np.mean((y_pred-y)**2)

def gradient_mse(X,y,w1,w0):
    y_pred = w1*X[:,0] + w0
    return np.array([2/len(X) * np.sum((y-y_pred)) *(-1),
                    2/len(X) * np.sum((y-y_pred) *(-X[:,0]))])

learning_rate = 0.1

next_value1 = 0
next_value0 = 0

for i in range(100):
    current_value1 = next_value1
    current_value0 = next_value0

    next_value1 = current_value1 - learning_rate*(gradient_mse(X,y,current_value1,current_value0)[1])
    next_value0 = current_value0 - learning_rate*(gradient_mse(X,y,current_value1,current_value0)[0])

    print(f'The iteration {i} is done')
    print(f'Last values were {current_value1} and {current_value0}')
    print(f'The new values are {next_value1} and {next_value0}')
    print(f'The mse is {mse(X,y,current_value1,current_value0)}')
    print("-"*30)


```

After 30 iterations it already found the right values, so i guess we need to make some constraints to avoid extra work

```python
def mse(X,y,w1,w0):
    y_pred = w1*X[:,0] + w0
    return np.mean((y_pred-y)**2)

def gradient_mse(X,y,w1,w0):
    y_pred = w1*X[:,0] + w0
    return np.array([2/len(X) * np.sum((y-y_pred)) *(-1),
                    2/len(X) * np.sum((y-y_pred) *(-X[:,0]))])

learning_rate = 0.1
const = 0.0001
next_value1 = 0
next_value0 = 0

for i in range(100):
    current_value1 = next_value1
    current_value0 = next_value0

    next_value1 = current_value1 - learning_rate*(gradient_mse(X,y,current_value1,current_value0)[1])
    next_value0 = current_value0 - learning_rate*(gradient_mse(X,y,current_value1,current_value0)[0])

    print(f'The iteration {i} is done')
    print(f'Last values were {current_value1} and {current_value0}')
    print(f'The new values are {next_value1} and {next_value0}')
    print(f'The mse is {mse(X,y,current_value1,current_value0)}')
    print("-"*30)

    if(abs(current_value1-next_value1)<const):
        break


```

We have more features, but writing the code manually "would become uncomfortable. As a solution we would use scalar multiplication. Also, writing the w0 would be little problematic, so we will add a new column of ones to the X

```python
#here we need to write it with the form of matrices
def mse_mat(X,y,w):
    y_pred = X@w
    return np.mean((y-y_pred)**2)

def gradient_mse_mat(X,y,w):
    y_pred = X@w
    return 2/len(X) * (y-y_pred) @ (-X)

#adding ones
X = np.column_stack([X, np.ones(len(X))])
```

```python
#initializing starting weights
next_weights = np.zeros(6)

for i in range(100):
    current_weights = next_weights

    next_weights = current_weights - learning_rate*gradient_mse_mat(X,y,current_weights)

    print(f'The iteration {i} is done')
    print(f'Last values were {current_weights}')
    print(f'The new values are {next_weights}')
    print(f'The mse is {mse_mat(X,y,current_weights)}')
    print("-"*30)

    if(any(next_weights-current_weights)<0.001):
        break



```
